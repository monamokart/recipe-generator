```python
import ast
import math

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from torch.optim.lr_scheduler import CosineAnnealingLR
```

## Encoders & feature tables

Build label encoders for ingredients and categories, normalise numerical features, and build multi-label vocabulary for cuisine and recipe-category heads.

```python
from collections import Counter

NUMERICAL_COLS = ["kcal", "protein_g", "carbs_g", "fat_g", "fiber_g", "sugar_g", "sodium_mg"]
NUM_NUMERICAL = len(NUMERICAL_COLS)

# ── Special tokens ────────────────────────────────────────────────────────────
SPECIAL_TOKENS = ["<PAD>", "<BOS>", "<CLS>"]
all_ingredients = SPECIAL_TOKENS + ingredient_df["ingredient"].tolist()
ingr_to_idx = {name: idx for idx, name in enumerate(all_ingredients)}
idx_to_ingr = {idx: name for name, idx in ingr_to_idx.items()}
PAD_IDX = ingr_to_idx["<PAD>"]
BOS_IDX = ingr_to_idx["<BOS>"]
CLS_IDX = ingr_to_idx["<CLS>"]
N_INGR = len(all_ingredients)
print(f"Ingredient vocab size (incl. specials): {N_INGR}")


# ── Ingredient-category label encoder ─────────────────────────────────────────
cat_vocab = ["<PAD>"] + sorted(ingredient_df["category"].unique().tolist())
cat_to_idx = {c: i for i, c in enumerate(cat_vocab)}
N_CAT = len(cat_vocab)
print(f"Category vocab size: {N_CAT}  →  {cat_vocab}")

# ── Lookup buffers (ingredient_idx → category_id / numerical) ─────────────────
ingr_cat_ids     = torch.zeros(N_INGR, dtype=torch.long)
ingr_numerical   = torch.zeros(N_INGR, NUM_NUMERICAL)

for _, row in ingredient_df.iterrows():
    idx = ingr_to_idx.get(row["ingredient"])
    if idx is not None:
        ingr_cat_ids[idx]   = cat_to_idx.get(str(row["category"]), 0)
        ingr_numerical[idx] = torch.tensor(
            [float(row[c]) for c in NUMERICAL_COLS], dtype=torch.float32
        )

# ── Standard-scale numerical features (statistics from real ingredients only) ─
real_mask = torch.tensor(
    [ingr_to_idx[n] not in range(len(SPECIAL_TOKENS)) for n in all_ingredients]
)
num_mean = ingr_numerical[real_mask].mean(0)
num_std  = ingr_numerical[real_mask].std(0).clamp(min=1e-6)
ingr_numerical_norm = (ingr_numerical - num_mean) / num_std
ingr_numerical_norm[:len(SPECIAL_TOKENS)] = 0.0   # zero-out special tokens

print(f"Numerical mean  : {num_mean.tolist()}")
print(f"Numerical std   : {num_std.tolist()}")

# ── Multi-label helpers ────────────────────────────────────────────────────────
def parse_str_list(s: str) -> list[str]:
    """Parse a Python-list string like \"['A', 'B']\" into a list of strings."""
    try:
        return [x.strip() for x in ast.literal_eval(s)]
    except Exception:
        return []
    
def get_top_labels(series: pd.Series, top_n: int = 20) -> set[str]:
    """Return the top_n most frequent labels across a Series of Python-list strings."""
    counter = Counter()
    for val in series.dropna():
        counter.update(parse_str_list(str(val)))
    return {label for label, _ in counter.most_common(top_n)}

def filter_multilabel(val: str, keep: set[str], other_label: str = "Other") -> str:
    """Keep only top labels in val; replace any others with other_label."""
    labels = parse_str_list(str(val))
    filtered = [l for l in labels if l in keep]
    if any(l not in keep for l in labels):
        filtered.append(other_label)
    return str(filtered)

all_cuisine  = pd.concat([recipes_train["Cuisine"],  recipes_test["Cuisine"]],  ignore_index=True)
all_category = pd.concat([recipes_train["Category"], recipes_test["Category"]], ignore_index=True)

top_cuisines   = get_top_labels(all_cuisine,  top_n=20)
top_categories = get_top_labels(all_category, top_n=20)

for df in [recipes_train, recipes_test]:
    df["Cuisine"]  = df["Cuisine"].apply(lambda v: filter_multilabel(str(v), top_cuisines,   "Other"))
    df["Category"] = df["Category"].apply(lambda v: filter_multilabel(str(v), top_categories, "Other"))

print(f"Top cuisines   ({len(top_cuisines)}):   {sorted(top_cuisines)}")
print(f"Top categories ({len(top_categories)}): {sorted(top_categories)}")

cuisine_vocab = sorted(top_cuisines) + ["Other"]
recipe_cat_vocab = sorted(top_categories) + ["Other"]

N_CUISINE = len(cuisine_vocab)
N_RECIPE_CAT = len(recipe_cat_vocab)


def encode_multilabel(val: str, vocab: list[str]) -> torch.Tensor:
    """Return a multi-hot float tensor for the labels present in *val*."""
    labels = set(parse_str_list(str(val)))
    return torch.tensor([1.0 if v in labels else 0.0 for v in vocab])
```

## Dataset & DataLoader

```python
class RecipeDataset(Dataset):
    """PyTorch Dataset for recipe ingredient sequences with multi-task labels.

    Each sample is a tuple of:
      - seq (list[int]): BOS + ingredient indices for one recipe.
      - cuisine_vec (Tensor[N_CUISINE]): multi-hot cuisine label.
      - cat_vec (Tensor[N_RECIPE_CAT]): multi-hot recipe-category label.

    Args:
        recipes_df: DataFrame with columns 'Ingredients', 'Cuisine', 'Category'.
    """

    def __init__(self, recipes_df: pd.DataFrame) -> None:
        self.records = []
        skipped = 0
        for _, row in recipes_df.iterrows():
            try:
                ingr_list: list[str] = ast.literal_eval(row["Ingredients"])
            except Exception:
                skipped += 1
                continue

            ingr_ids = [ingr_to_idx.get(i, PAD_IDX) for i in ingr_list]
            # Require ≥2 ingredients and no unknown (PAD) entries
            if len(ingr_ids) < 2 or PAD_IDX in ingr_ids:
                skipped += 1
                continue

            seq = [BOS_IDX] + ingr_ids
            cuisine_vec = encode_multilabel(str(row.get("Cuisine", "[]")), cuisine_vocab)
            cat_vec = encode_multilabel(str(row.get("Category", "[]")), recipe_cat_vocab)
            self.records.append((seq, cuisine_vec, cat_vec))

        print(f"Loaded {len(self.records)} recipes ({skipped} skipped).")

    def __len__(self) -> int:
        return len(self.records)

    def __getitem__(self, idx: int) -> tuple[list[int], torch.Tensor, torch.Tensor]:
        return self.records[idx]


def collate_fn(
    batch: list[tuple[list[int], torch.Tensor, torch.Tensor]],
) -> tuple[torch.Tensor, torch.Tensor, torch.Tensor, torch.Tensor, torch.Tensor]:
    """Pad a batch of variable-length sequences and append the CLS token.

    CLS is inserted immediately after the last real ingredient so its position
    is tracked per-sample. The sequence in memory looks like:

        [BOS, i_1, ..., i_n, CLS, PAD, ..., PAD]

    Args:
        batch: List of (seq, cuisine_vec, cat_vec) tuples.

    Returns:
        Tuple of:
          - ingr_ids:     long  (B, T_max) – padded sequences with CLS embedded.
          - pad_mask:     bool  (B, T_max) – True at PAD positions.
          - cls_positions: long (B,)       – index of CLS token in each row.
          - cuisine_vecs: float (B, N_CUISINE).
          - cat_vecs:     float (B, N_RECIPE_CAT).
    """
    seqs, cuisine_vecs, cat_vecs = zip(*batch)

    # Append CLS to every sequence before padding
    seqs_cls = [s + [CLS_IDX] for s in seqs]
    max_len = max(len(s) for s in seqs_cls)
    B = len(seqs_cls)

    ingr_ids      = torch.full((B, max_len), PAD_IDX, dtype=torch.long)
    pad_mask      = torch.ones((B, max_len), dtype=torch.bool)   # True = ignore
    cls_positions = torch.zeros(B, dtype=torch.long)

    for i, s in enumerate(seqs_cls):
        ingr_ids[i, :len(s)] = torch.tensor(s, dtype=torch.long)
        pad_mask[i, :len(s)] = False
        cls_positions[i]     = len(s) - 1   # CLS is always the last real token

    return ingr_ids, pad_mask, cls_positions, torch.stack(cuisine_vecs), torch.stack(cat_vecs)


train_dataset = RecipeDataset(recipes_train)
test_dataset  = RecipeDataset(recipes_test)
print(f"Train: {len(train_dataset)}  |  Test: {len(test_dataset)}")
```

## Model Architecture

`RecipeTransformer` is a **causal transformer encoder** with three output heads:

| Head | Input | Target |
|---|---|---|
| **AR** | every token position | next ingredient token |
| **Cuisine** | CLS output | multi-hot cuisine vector |
| **Category** | CLS output | multi-hot recipe-category vector |

Each token's input embedding is the concatenation of:
- learned ingredient-name embedding (`d_ingr_emb`)
- learned ingredient-category embedding (`d_cat_emb`)
- normalised nutritional feature vector (7 values)

→ projected to `d_model` before entering the transformer stack.


```python
# Derived device
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {DEVICE}")
```

```python
import math


class RecipeTransformer(nn.Module):
    """Causal transformer for multi-task recipe modelling.

    Three simultaneous output heads:
      1. **AR head**: predict the next ingredient at every sequence position
         (cross-entropy, causal mask, CLS and PAD positions excluded).
      2. **Cuisine head**: multi-label prediction via the CLS token output.
      3. **Category head**: multi-label prediction via the CLS token output.

    Token input vector = ingredient-name embedding ⊕ ingredient-category
    embedding ⊕ normalised nutritional vector, projected to ``d_model``.

    Args:
        config:        Architecture / training configuration dictionary.
        n_ingr:        Ingredient vocab size (includes specials + CLS).
        n_cat:         Number of ingredient-category labels.
        n_cuisine:     Number of recipe cuisine labels.
        n_recipe_cat:  Number of recipe category labels.
        num_numerical: Length of per-ingredient nutritional feature vector.
        ingr_cat_buf:  LongTensor (n_ingr,)            – idx → category idx.
        ingr_num_buf:  FloatTensor (n_ingr, num_numerical) – normalised features.
        pad_idx:       PAD index in ingredient vocabulary.
        cat_pad_idx:   PAD index in ingredient-category vocabulary.
    """

    def __init__(
        self,
        config: dict,
        n_ingr: int,
        n_cat: int,
        n_cuisine: int,
        n_recipe_cat: int,
        num_numerical: int,
        ingr_cat_buf: torch.Tensor,
        ingr_num_buf: torch.Tensor,
        pad_idx: int,
        cat_pad_idx: int = 0,
    ) -> None:
        super().__init__()

        d_ingr  = config["d_ingr_emb"]
        d_cat   = config["d_cat_emb"]
        d_model = config["d_model"]

        # ── Non-trainable lookup buffers ──────────────────────────────────────
        self.register_buffer("ingr_cat_buf", ingr_cat_buf)
        self.register_buffer("ingr_num_buf", ingr_num_buf)

        # ── Embeddings ────────────────────────────────────────────────────────
        self.ingr_emb = nn.Embedding(n_ingr, d_ingr, padding_idx=pad_idx)
        self.cat_emb  = nn.Embedding(n_cat,  d_cat,  padding_idx=cat_pad_idx)

        # ── Input projection: concat → d_model ────────────────────────────────
        in_dim = d_ingr + d_cat + num_numerical
        self.input_proj  = nn.Linear(in_dim, d_model)
        self.dropout_in  = nn.Dropout(config["dropout"])

        # ── Transformer (Pre-LayerNorm for stable training) ───────────────────
        enc_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=config["n_heads"],
            dim_feedforward=config["d_ff"],
            dropout=config["dropout"],
            batch_first=True,
            norm_first=True,   # Pre-LN
        )
        self.transformer = nn.TransformerEncoder(
            enc_layer,
            num_layers=config["n_layers"],
            enable_nested_tensor=False,
        )

        # ── Output heads ──────────────────────────────────────────────────────
        self.ar_head      = nn.Linear(d_model, n_ingr)
        self.cuisine_head = nn.Linear(d_model, n_cuisine)
        self.cat_head     = nn.Linear(d_model, n_recipe_cat)

        self._init_weights()

    # ── Weight initialisation ─────────────────────────────────────────────────
    def _init_weights(self) -> None:
        for emb in (self.ingr_emb, self.cat_emb):
            nn.init.trunc_normal_(emb.weight, std=0.02)
        for head in (self.ar_head, self.cuisine_head, self.cat_head, self.input_proj):
            nn.init.trunc_normal_(head.weight, std=0.02)
            nn.init.zeros_(head.bias)

    # ── Forward pass ─────────────────────────────────────────────────────────
    def forward(
        self,
        ingr_ids: torch.Tensor,       # (B, T)
        pad_mask: torch.Tensor,        # (B, T)  True = PAD position
        cls_positions: torch.Tensor,   # (B,)    index of CLS in each row
    ) -> tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
        """Run a forward pass.

        Returns:
            ar_logits:      (B, T, n_ingr)   – logits for next-token prediction.
            cuisine_logits: (B, n_cuisine)   – raw scores for cuisine labels.
            cat_logits:     (B, n_recipe_cat)– raw scores for category labels.
        """
        B, T = ingr_ids.shape
        device = ingr_ids.device

        # Per-token feature vector ─────────────────────────────────────────────
        name_emb = self.ingr_emb(ingr_ids)              # (B, T, d_ingr)
        cat_ids  = self.ingr_cat_buf[ingr_ids]           # (B, T)
        cat_emb  = self.cat_emb(cat_ids)                 # (B, T, d_cat)
        num_feat = self.ingr_num_buf[ingr_ids]           # (B, T, num_numerical)

        feat = torch.cat([name_emb, cat_emb, num_feat], dim=-1)  # (B, T, in_dim)
        feat = self.dropout_in(
            self.input_proj(feat)                        # (B, T, d_model)
        )

        # Causal mask (bool) – same dtype as src_key_padding_mask ─────────────
        causal_mask = torch.triu(
            torch.ones((T, T), device=device, dtype=torch.bool),
            diagonal=1,
        )

        out = self.transformer(
            feat,
            mask=causal_mask,
            src_key_padding_mask=pad_mask,
            is_causal=True,
        )  # (B, T, d_model)

        # AR logits at every position ──────────────────────────────────────────
        ar_logits = self.ar_head(out)  # (B, T, n_ingr)

        # CLS representation – gather per-sample ──────────────────────────────
        cls_idx = cls_positions.view(B, 1, 1).expand(B, 1, out.size(-1))
        cls_out = out.gather(1, cls_idx).squeeze(1)   # (B, d_model)

        cuisine_logits = self.cuisine_head(cls_out)   # (B, n_cuisine)
        cat_logits     = self.cat_head(cls_out)       # (B, n_recipe_cat)

        return ar_logits, cuisine_logits, cat_logits


def build_model(config: dict) -> RecipeTransformer:
    """Instantiate and return a ``RecipeTransformer`` from *config*."""
    model = RecipeTransformer(
        config=config,
        n_ingr=N_INGR,
        n_cat=N_CAT,
        n_cuisine=N_CUISINE,
        n_recipe_cat=N_RECIPE_CAT,
        num_numerical=NUM_NUMERICAL,
        ingr_cat_buf=ingr_cat_ids.clone(),
        ingr_num_buf=ingr_numerical_norm.clone(),
        pad_idx=PAD_IDX,
        cat_pad_idx=0,
    ).to(DEVICE)

    n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"RecipeTransformer  –  trainable parameters: {n_params:,}")
    return model
```

## Training

The `train` function orchestrates:
- **3 losses** computed every batch: AR cross-entropy, cuisine BCE, category BCE (combined with config weights).
- **Metrics tracked each epoch**: AR loss + accuracy, cuisine/category loss + macro F1.
- **Early stopping** on total validation loss (patience from config).
- **History dict** returned with all per-epoch train/val metrics.


```python
import copy
from sklearn.metrics import f1_score as sk_f1_score


# ── Loss helpers ──────────────────────────────────────────────────────────────

def ar_loss_and_acc(
    ar_logits: torch.Tensor,      # (B, T, V)
    ingr_ids: torch.Tensor,       # (B, T)
    pad_mask: torch.Tensor,       # (B, T)  True = PAD
    cls_positions: torch.Tensor,  # (B,)
) -> tuple[torch.Tensor, float]:
    """AR cross-entropy and top-1 accuracy, excluding CLS and PAD positions.

    Position *t* predicts token at position *t+1*.  We include positions
    ``0 .. cls_pos[i] - 2`` for sample *i* (i.e. we predict all real
    ingredients but stop before predicting the CLS token itself).
    """
    B, T, V = ar_logits.shape
    device = ingr_ids.device

    pos_idx = torch.arange(T, device=device).unsqueeze(0).expand(B, T)  # (B, T)
    cls_pos = cls_positions.unsqueeze(1)                                  # (B, 1)

    # valid output positions: not PAD, not the last real-ingredient position
    # (which would predict CLS), and not CLS itself
    valid = (pos_idx < cls_pos - 1) & (~pad_mask)   # (B, T)

    # Shift targets left: target[i,t] = ingr_ids[i,t+1]
    targets = torch.roll(ingr_ids, -1, dims=1)       # (B, T) – wrap-around irrelevant
    targets[:, -1] = PAD_IDX                         # sentinel for last column

    logits_flat  = ar_logits[valid]    # (N_valid, V)
    targets_flat = targets[valid]      # (N_valid,)

    if logits_flat.size(0) == 0:
        return ar_logits.sum() * 0.0, 0.0

    loss = F.cross_entropy(logits_flat, targets_flat)
    with torch.no_grad():
        acc = (logits_flat.argmax(-1) == targets_flat).float().mean().item()
    return loss, acc


def cls_loss_and_f1(
    logits: torch.Tensor,   # (B, n_labels)
    targets: torch.Tensor,  # (B, n_labels)  multi-hot float
    threshold: float = 0.5,
) -> tuple[torch.Tensor, float]:
    """Binary cross-entropy and macro multi-label F1."""
    loss = F.binary_cross_entropy_with_logits(logits, targets)
    with torch.no_grad():
        preds = (torch.sigmoid(logits) >= threshold).cpu().numpy().astype(int)
        tgts  = targets.cpu().numpy().astype(int)
        # Zero-division set to 0.0; sklearn handles all-zero rows gracefully
        f1 = sk_f1_score(tgts, preds, average="macro", zero_division=0)
    return loss, float(f1)


# ── Single epoch helpers ──────────────────────────────────────────────────────

def _run_epoch(
    model: RecipeTransformer,
    loader: DataLoader,
    config: dict,
    optimizer: torch.optim.Optimizer | None = None,
) -> dict[str, float]:
    """Run one training (optimizer != None) or evaluation epoch.

    Returns a dict with keys:
        ar_loss, ar_acc, cuisine_loss, cuisine_f1, cat_loss, cat_f1, total_loss
    """
    training = optimizer is not None
    model.train(training)

    w_ar  = config["w_ar"]
    w_cus = config["w_cuisine"]
    w_cat = config["w_category"]

    totals: dict[str, float] = {k: 0.0 for k in
        ["ar_loss", "ar_acc", "cuisine_loss", "cuisine_f1", "cat_loss", "cat_f1", "total_loss"]}
    n_batches = 0

    ctx = torch.enable_grad() if training else torch.no_grad()
    with ctx:
        for ingr_ids, pad_mask, cls_positions, cuisine_vecs, cat_vecs in loader:
            ingr_ids = ingr_ids.to(DEVICE)
            pad_mask = pad_mask.to(DEVICE)
            cls_positions = cls_positions.to(DEVICE)
            cuisine_vecs = cuisine_vecs.to(DEVICE)
            cat_vecs = cat_vecs.to(DEVICE)

            ar_logits, cuisine_logits, cat_logits = model(ingr_ids, pad_mask, cls_positions)

            ar_l,  ar_a   = ar_loss_and_acc(ar_logits, ingr_ids, pad_mask, cls_positions)
            cus_l, cus_f1 = cls_loss_and_f1(cuisine_logits, cuisine_vecs)
            cat_l, cat_f1 = cls_loss_and_f1(cat_logits,     cat_vecs)

            total = w_ar * ar_l + w_cus * cus_l + w_cat * cat_l

            if training:
                optimizer.zero_grad()
                total.backward()
                nn.utils.clip_grad_norm_(model.parameters(), config["grad_clip"])
                optimizer.step()

            totals["ar_loss"] += ar_l.item()
            totals["ar_acc"] += ar_a
            totals["cuisine_loss"] += cus_l.item()
            totals["cuisine_f1"] += cus_f1
            totals["cat_loss"] += cat_l.item()
            totals["cat_f1"] += cat_f1
            totals["total_loss"] += total.item()
            n_batches += 1

    return {k: v / max(n_batches, 1) for k, v in totals.items()}


# ── Main training function ────────────────────────────────────────────────────

def train(
    config: dict,
    train_dataset: RecipeDataset,
    val_dataset: RecipeDataset,
) -> tuple[RecipeTransformer, dict[str, list[float]]]:
    """Train a RecipeTransformer and return (best_model, history).

    Args:
        config:        Configuration dictionary (see CONFIG cell).
        train_dataset: Training split RecipeDataset.
        val_dataset:   Validation split RecipeDataset.

    Returns:
        best_model: Model weights restored to the best validation checkpoint.
        history:    Dict mapping metric names to per-epoch lists.
                    Keys: train_*/val_* for each of
                    ar_loss, ar_acc, cuisine_loss, cuisine_f1,
                    cat_loss, cat_f1, total_loss.
    """
    train_loader = DataLoader(
        train_dataset,
        batch_size=config["batch_size"],
        shuffle=True,
        collate_fn=collate_fn,
        drop_last=False,
    )
    val_loader = DataLoader(
        val_dataset,
        batch_size=config["batch_size"],
        shuffle=False,
        collate_fn=collate_fn,
    )

    model = build_model(config)
    optimizer = AdamW(model.parameters(), lr=config["lr"], weight_decay=config["weight_decay"])
    scheduler = CosineAnnealingLR(optimizer, T_max=config["epochs"])

    history: dict[str, list[float]] = {
        f"{split}_{metric}": []
        for split  in ("train", "val")
        for metric in ("ar_loss", "ar_acc", "cuisine_loss", "cuisine_f1",
                       "cat_loss", "cat_f1",  "total_loss")
    }

    patience        = config["early_stopping_patience"]
    best_val_loss   = float("inf")
    no_improve      = 0
    best_state_dict = copy.deepcopy(model.state_dict())

    for epoch in range(1, config["epochs"] + 1):
        train_metrics = _run_epoch(model, train_loader, config, optimizer=optimizer)
        val_metrics   = _run_epoch(model, val_loader,   config, optimizer=None)
        scheduler.step()

        # Record history
        for metric, val in train_metrics.items():
            history[f"train_{metric}"].append(val)
        for metric, val in val_metrics.items():
            history[f"val_{metric}"].append(val)

        print(
            f"Epoch {epoch:02d}/{config['epochs']}  "
            f"| train  loss={train_metrics['total_loss']:.4f}  "
            f"ar_acc={train_metrics['ar_acc']:.3f}  "
            f"cus_f1={train_metrics['cuisine_f1']:.3f}  "
            f"cat_f1={train_metrics['cat_f1']:.3f}  "
            f"| val  loss={val_metrics['total_loss']:.4f}  "
            f"ar_acc={val_metrics['ar_acc']:.3f}  "
            f"cus_f1={val_metrics['cuisine_f1']:.3f}  "
            f"cat_f1={val_metrics['cat_f1']:.3f}"
        )

        # Early stopping
        if val_metrics["total_loss"] < best_val_loss:
            best_val_loss   = val_metrics["total_loss"]
            best_state_dict = copy.deepcopy(model.state_dict())
            no_improve      = 0
        else:
            no_improve += 1
            if no_improve >= patience:
                print(f"Early stopping triggered after epoch {epoch} "
                      f"(no improvement for {patience} epochs).")
                break

    # Restore best weights
    model.load_state_dict(best_state_dict)
    print(f"\nTraining complete. Best val loss: {best_val_loss:.4f}")
    return model, history
```

```python
import datetime
import json
import os
import tempfile

from google.cloud import storage


# ── GCS save utility ──────────────────────────────────────────────────────────

def save_run_to_gcs(
    model: RecipeTransformer,
    config: dict,
    history: dict[str, list[float]],
    vocab_meta: dict,
) -> str:
    """Save model weights, config, history, and vocab metadata to GCS.

    Files are uploaded under:
        gs://<bucket>/<prefix>/<YYYYMMDD_HHMMSS>/
            model.pt       – PyTorch state dict
            config.json    – training configuration
            history.json   – per-epoch metrics
            vocab_meta.json– vocab sizes and special-token indices

    Args:
        model:      Trained RecipeTransformer (best weights loaded).
        config:     CONFIG dict used for this run.
        history:    History dict returned by ``train()``.
        vocab_meta: Dict of vocabulary sizes and special token indices.

    Returns:
        GCS path prefix of the saved run, e.g.
        ``gs://recipe-generation/models/20260319_143022``.
    """
    timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    run_prefix = f"{config['gcs_model_prefix']}/{timestamp}"

    gcs_client = storage.Client()
    bucket     = gcs_client.bucket(config["gcs_bucket"])

    # ── Model weights ─────────────────────────────────────────────────────────
    with tempfile.NamedTemporaryFile(suffix=".pt", delete=False) as f:
        tmp_path = f.name
    torch.save(model.state_dict(), tmp_path)
    bucket.blob(f"{run_prefix}/model.pt").upload_from_filename(tmp_path)
    os.unlink(tmp_path)

    # ── Config ────────────────────────────────────────────────────────────────
    bucket.blob(f"{run_prefix}/config.json").upload_from_string(
        json.dumps(config, indent=2), content_type="application/json"
    )

    # ── History ───────────────────────────────────────────────────────────────
    history_serializable = {k: [float(v) for v in vals] for k, vals in history.items()}
    bucket.blob(f"{run_prefix}/history.json").upload_from_string(
        json.dumps(history_serializable, indent=2), content_type="application/json"
    )

    # ── Vocab metadata ────────────────────────────────────────────────────────
    bucket.blob(f"{run_prefix}/vocab_meta.json").upload_from_string(
        json.dumps(vocab_meta, indent=2), content_type="application/json"
    )

    gcs_path = f"gs://{config['gcs_bucket']}/{run_prefix}"
    print(f"Run saved to: {gcs_path}")
    return gcs_path
```
