```python
import pandas as pd
from ast import literal_eval
```

```python
# ! gcloud auth application-default login
recipes = pd.read_csv('gs://recipe-generation/recipes.csv', sep=";")
nutri_data = pd.read_csv('gs://recipe-generation/nutri_data.csv')
```

```python
recipes = recipes.dropna(subset=['Ingredients'])
ingredients_count = recipes['Ingredients'].apply(lambda l : [val["ingredient"] for val in literal_eval(l)]).explode().value_counts()
ingredients_count
```

```python
top_ingredients = ingredients_count.head(500)
top_ingredients
```

```python
filtered_recipes = recipes[recipes['Ingredients'].apply(
    lambda l: all(val["ingredient"] in top_ingredients.index for val in literal_eval(l))
)]
filtered_recipes
```

```python
from transformers import pipeline
import numpy as np

def cosine_similarity(query_embedding, candidate_embeddings):
    query_vec = np.array(query_embedding)
    candidate_vecs = np.array(candidate_embeddings)
    # Normalize vectors
    query_norm = query_vec / np.linalg.norm(query_vec)
    candidate_norms = candidate_vecs / np.linalg.norm(candidate_vecs, axis=1, keepdims=True)
    # Compute cosine similarity
    similarities = np.dot(candidate_norms, query_norm)
    return similarities

def ingredient_retrieval(query, candidates):
    # Use a sentence-transformers model for semantic search
    retrieval = pipeline("feature-extraction", model="sentence-transformers/all-MiniLM-L6-v2")
    # Encode query and candidates
    query_embedding = retrieval(query)[0][0]
    candidate_embeddings = [retrieval(c)[0][0] for c in candidates]
    # Compute cosine similarity
    similarities = cosine_similarity(query_embedding, candidate_embeddings)
    # Return best match and scores
    best_idx = similarities.argmax()
    return candidates[best_idx], similarities[best_idx]

# Example usage:
query = top_ingredients.index[0]
candidates = nutri_data["Shrt_Desc"].tolist()
best_match, score = ingredient_retrieval(query, candidates)
```
